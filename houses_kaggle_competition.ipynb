## Baseline Pipeline

We will now build a **full baseline pipeline** step by step.

The goal is to create a clean `Pipeline` that can take the raw features `X`, preprocess them, and train a baseline model in one workflow.

Our final pipeline will combine:

- a preprocessing pipeline for numerical features
- a preprocessing pipeline for categorical features
- a `ColumnTransformer` to apply each preprocessing block to the right columns
- a simple regression model

We will build each components step by step:

1. **Numerical preprocessing**
   - Select numerical columns
   - Impute missing values
   - Scale numerical features

2. **Categorical preprocessing**
   - Select categorical columns
   - Impute missing values
   - Encode categorical features

3. **Full preprocessing block with ColumnTransformer**
   - Apply the numerical pipeline to numerical columns
   - Apply the categorical pipeline to categorical columns

4. **Full baseline pipeline**
   - Combine preprocessing with a regression model
   - Fit the full pipeline on the training data
